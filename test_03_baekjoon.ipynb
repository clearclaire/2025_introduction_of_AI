```python
# 10250
import numpy as np

a = int(input())
for _ in range(a):
    h, w, n = map(int, input().split())
    # y 층번호 x 호실번호
    if n % h == 0:
        y = h
        x = n // h
    else:
        y = n % h
        x = n // h + 1
    print(f"{y}{x:02d}")
```
